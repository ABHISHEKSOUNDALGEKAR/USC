```python
# Libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.io import arff
from sklearn import metrics
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, f1_score, precision_score
```

1. Vertebral Column Data Set
This Biomedical data set was built by Dr. Henrique da Mota during a medical residence
period in Lyon, France. Each patient in the data set is represented in the data set
by six biomechanical attributes derived from the shape and orientation of the pelvis
and lumbar spine (in this order): pelvic incidence, pelvic tilt, lumbar lordosis angle,
sacral slope, pelvic radius and grade of spondylolisthesis. The following convention is
used for the class labels: DH (Disk Hernia), Spondylolisthesis (SL), Normal (NO) and
Abnormal (AB). In this exercise, we only focus on a binary classification task NO=0
and AB=1.

1.(a) Download the Vertebral Column Data Set from: https://archive.ics.uci.edu/ml/datasets/Vertebral+Column.

```python
try:
    data, meta = arff.loadarff('../data/vertebral_column_data/column_2C_weka.arff')
    df = pd.DataFrame(data)
    
    #columns loaded as bytes, decode them to str:
    for col in df.select_dtypes([object]):
        try:
            df[col] = df[col].str.decode('utf-8')
        except AttributeError:
            # Column wasn’t bytes no decoding
            pass
    
    # class labels to binary mapping.
    df['class'] = df['class'].map({'Normal': 0, 'Abnormal': 1})
    
except FileNotFoundError as fnf_err:
    print(f"File not found: {fnf_err}")
    df = None
except UnicodeDecodeError as uni_err:
    print(f"Decoding error: {uni_err}")
    df = None
except Exception as e:
    print(f"An unexpected error occurred: {e}")
    df = None

df
```

## 1.(c) Classification using KNN on Vertebral Column Data Set ##

### 1. c) i. Write code for k-nearest neighbors with Euclidean metric (or use a software package). ###

```python
X_train, y_train = training_set[var_indep], training_set['class']
X_test,  y_test  = test_set[var_indep],  test_set['class']

# KNN is fit with Euclidean distance (default p=2)
#knn = KNeighborsClassifier(n_neighbors=5)
knn = KNeighborsClassifier(n_neighbors=5, metric='euclidean')

knn.fit(X_train, y_train)

# Predict and report accuracy
y_pred = knn.predict(X_test)
print(f"Accuracy of the KNN model: {accuracy_score(y_test, y_pred):.2f}")
```

### 1. c) ii. Test all the data in the test database with k nearest neighbors. Take decisions by majority polling. Plot train and test errors in terms of k for k ∈ {208, 205, . . . , 7, 4, 1, } (in reverse order). You are welcome to use smaller increments of k. Which k∗ is the most suitable k among those values? Calculate the confusion matrix, true positive rate, true negative rate, precision, and F1-score when k = k∗. ###

```python
k_choices = np.arange(208, 0, -3)
errors_train = []
errors_test = []

for k_val in k_choices:   # Loop over different k values
    clf = KNeighborsClassifier(n_neighbors=k_val)
    clf.fit(X_train, y_train)
    
    # Compute error = 1 - accuracy
    errors_train.append(1 - clf.score(X_train, y_train))
    errors_test.append(1 - clf.score(X_test, y_test))

# training and testing error curves plots
fig1, ax1 = plt.subplots(figsize=(8, 4))
ax1.plot(k_choices, errors_train, marker='o', linestyle='-', color='green', label='Train Error')
ax1.plot(k_choices, errors_test,  marker='s', linestyle='--', color='red',   label='Test Error')
ax1.invert_xaxis()  # so that larger k appears on left
ax1.set_xlabel('Number of Neighbors (k)', fontsize=12)
ax1.set_ylabel('Error Rate', fontsize=12)
ax1.set_title('K vs. Error Rates', fontsize=14)
ax1.legend()
plt.tight_layout()
plt.show()

# k that yields the lowest test error
best_index = np.argmin(errors_test)
best_k = k_choices[best_index]
print(f"Selected k*: {best_k}")

# best_k and evaluate
best_clf = KNeighborsClassifier(n_neighbors=best_k)
best_clf.fit(X_train, y_train)
y_predicted = best_clf.predict(X_test)

# Confusion matrix and derived metrics
tn, fp, fn, tp = confusion_matrix(y_test, y_predicted).ravel()
tpr = tp / (tp + fn) if (tp + fn) > 0 else 0  # Sensitivity
tnr = tn / (tn + fp) if (tn + fp) > 0 else 0  # Specificity
precision = tp / (tp + fp) if (tp + fp) > 0 else 0
f1 = f1_score(y_test, y_predicted)

print("Confusion Matrix:")
print(np.array([[tn, fp],
                [fn, tp]]))
print(f"Sensitivity (TPR): {tpr:.4f}")
print(f"Specificity (TNR): {tnr:.4f}")
print(f"Precision: {precision:.4f}")
print(f"F1 Score: {f1:.4f}")

# Normalized confusion matrix as a heatmap
conf_mat = confusion_matrix(y_test, y_predicted)
conf_mat_norm = conf_mat / conf_mat.sum()

fig2, ax2 = plt.subplots(figsize=(6, 4))
sns.heatmap(
    conf_mat_norm,
    annot=True,
    fmt='.2%',
    cmap='Blues',
    cbar=False,
    ax=ax2
)
ax2.set_title('Normalized Confusion Matrix', fontsize=14)
ax2.set_xlabel('Predicted Label')
ax2.set_ylabel('True Label')
plt.tight_layout()
plt.show()
```

### 1. c) iii. Since the computation time depends on the size of the training set, one may only use a subset of the training set. Plot the best test error rate, 3 which is obtained by some value of k, against the size of training set, when the size of training set is N ∈ {10, 20, 30, . . . , 210}. 4 Note: for each N, select your training set by choosing the first b N/3c rows of Class 0 and the first N − bN/3c rows of Class 1 in the training set you created in 1(b)iii. Also, for each N, select the optimal k from a set starting from k = 1, increasing by 5.For example, if N = 200, the optimal k is selected from {1, 6, 11, . . . , 196}. This plot is called a Learning Curve. Let us further explore some variants of KNN. ###

```python

neg_samples = df[df['class'] == 0]     # Pre-split DataFrame by class
pos_samples = df[df['class'] == 1]

features = var_indep    # List of independent feature names

sample_sizes = np.arange(10, 211, 10)    # Arrays to collect results
best_errors = []
best_ks = []

for size in sample_sizes:
    count_neg = size // 3
    count_pos = size - count_neg
    small_df = pd.concat([neg_samples.iloc[:count_neg], pos_samples.iloc[:count_pos]])
    X_small = small_df[features]
    y_small = small_df['class']
    k_list = np.arange(1, size, 5)
    test_err_list = []

    for k_val in k_list:
        model = KNeighborsClassifier(n_neighbors=k_val).fit(X_small, y_small)
        test_err_list.append(1 - model.score(X_test, y_test))

    idx_min = np.argmin(test_err_list)
    best_errors.append(test_err_list[idx_min])
    best_ks.append(int(k_list[idx_min]))

plt.figure(figsize=(10, 6))    # plotting begins
plt.plot(sample_sizes, best_errors, 'o-', color='teal', label='Best Test Error')

# Annotate each point with its test-error value
for x, y in zip(sample_sizes, best_errors):
    plt.text(x, y, f"{y:.3f}", ha='center', va='bottom', fontsize= 9)

plt.title('Lowest Test Error vs. Training Size', fontsize=14)
plt.xlabel('Training Sample Size', fontsize=12)
plt.ylabel('Test Error', fontsize=12)
plt.xticks(sample_sizes)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
k_candidates = np.arange(1, 200, 5)
error_dict = {}

for k in k_candidates:
    model = KNeighborsClassifier(n_neighbors=k, p=1).fit(X_train, y_train)
    error_dict[int(k)] = 1 - model.score(X_test, y_test)

min_error = min(error_dict.values())

# sarey k’s that achieve that minimum error
optimal_ks = [k for k, err in error_dict.items() if err == min_error]

print("Test errors by k:")
print(error_dict)
print("\nMinimum test error:", min_error)
print("All k values achieving this minimum error:", optimal_ks)
```

### 1. d) i. C. which becomes Chebyshev Distance with p → ∞ ###

```python
k_values = range(1, 197, 5)

errors_ch = [
    1 - KNeighborsClassifier(n_neighbors=k, metric='chebyshev')
        .fit(X_train, y_train)
        .score(X_test, y_test)
    for k in k_values
]

#smallest Chebyshev error
chebyshev_min_err = min(errors_ch)

#all k’s that achieve this minimum error
chebyshev_all_best_ks = [
    k for k, err in zip(k_values, errors_ch)
    if err == chebyshev_min_err
]

print(f"All k’s achieving minimum Chebyshev error ({chebyshev_min_err:.4f}): {chebyshev_all_best_ks}")
print(f"Test error (Chebyshev): {chebyshev_min_err:.4f}")
```

## 1. d) ii. Mahalanobis Distance. ##

```python
cov_mat = np.cov(X_train, rowvar=False)
try:
    cov_inv = np.linalg.inv(cov_mat)
except np.linalg.LinAlgError:
    cov_inv = np.linalg.pinv(cov_mat)

errors_maha = [
    1 - KNeighborsClassifier(
            n_neighbors=k,
            metric='mahalanobis',
            metric_params={'VI': cov_inv}
        ).fit(X_train, y_train).score(X_test, y_test)
    for k in k_values
]

best_idx = np.argmin(errors_maha)    # Index of minimal test error
best_k_maha = k_values[best_idx]
best_error_maha = errors_maha[best_idx]

best_maha = best_k_maha

print(f"Mahalanobis optimal k: {best_k_maha}")
print(f"Test error (Mahalanobis): {best_error_maha:.6f}")

```

## 1.(e) The majority polling decision can be replaced by weighted decision, in which the weight of each point in voting is inversely proportional to its distance from the query/test data point. In this case, closer neighbors of a query point will have a greater influence than neighbors which are further away. Use weighted voting with Euclidean, Manhattan, and Chebyshev distances and report the best test errors when k ∈ {1, 6, 11, 16, . . . , 196}. ##

```python
features = var_indep
X_tr, y_tr = training_set[features], training_set['class']
X_te, y_te = test_set[features],  test_set['class']

distance_metrics = ['euclidean', 'manhattan', 'chebyshev']
k_candidates = np.arange(1, 197, 5)

outcomes = {}
for dist in distance_metrics:
    test_errs = [
        1 - KNeighborsClassifier(
                n_neighbors=k,
                weights='distance',
                metric=dist
            ).fit(X_tr, y_tr).score(X_te, y_te)
        for k in k_candidates
    ]
    best_idx = np.argmin(test_errs)
    outcomes[dist] = (int(k_candidates[best_idx]), test_errs[best_idx])

summary_df = pd.DataFrame([
    {"Distance": dist.capitalize(), "Best k": k_opt, "Error": err_opt}
    for dist, (k_opt, err_opt) in outcomes.items()
])

table_styles = [
    {"selector": "th", "props": [("border", "1px solid black"), ("background-color", "#f2f2f2")]},
    {"selector": "td", "props": [("border", "1px solid black")]},
    {"selector": "table", "props": [("border-collapse", "collapse"), ("width", "50%")]}
]
styled_summary_df = summary_df.style.set_table_styles(table_styles).set_properties(**{"text-align": "center"})
styled_summary_df
```

### 0% (zero training error), which occurs when using k=1 ###

```python
cov = np.cov(X_train, rowvar=False)
try:
    cov_inv = np.linalg.inv(cov)
except np.linalg.LinAlgError:
    cov_inv = np.linalg.pinv(cov)

metrics_list = ['euclidean', 'manhattan', 'chebyshev', 'mahalanobis']
k_values_to_check = [1, 6, 11, 26]

print("Training error for k ∈ {1, 6, 11, 26} under each metric:")
for dist in metrics_list:
    line = [f"{dist.capitalize():<10}"]
    for k in k_values_to_check:
        if dist == 'mahalanobis':
            clf = KNeighborsClassifier(n_neighbors=k, metric='mahalanobis', metric_params={'VI': cov_inv})
        else:
            clf = KNeighborsClassifier(n_neighbors=k, metric=dist)
        clf.fit(X_train, y_train)
        tr_err = 1 - clf.score(X_train, y_train)
        line.append(f"k={k}: {tr_err:.3%}")
    print("  " + " | ".join(line))
```
