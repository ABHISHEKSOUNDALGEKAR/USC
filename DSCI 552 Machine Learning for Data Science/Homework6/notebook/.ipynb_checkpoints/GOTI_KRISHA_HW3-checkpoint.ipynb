### (b) Keep datasets 1 and 2 in folders bending1 and bending 2, as well as datasets 1, 2, and 3 in other folders as test data and other datasets as train data

```python

# File structure mapping
file_structure = {
    'bending1': {'test': [1, 2], 'train': range(3, 8)},
    'bending2': {'test': [1, 2], 'train': range(3, 7)},
    'cycling': {'test': [1, 2, 3], 'train': range(4, 16)},
    'lying': {'test': [1, 2, 3], 'train': range(4, 16)},
    'sitting': {'test': [1, 2, 3], 'train': range(4, 16)},
    'standing': {'test': [1, 2, 3], 'train': range(4, 16)},
    'walking': {'test': [1, 2, 3], 'train': range(4, 16)}
}

test_data, test_labels = [], []
train_data, train_labels = [], []

activity = ['bending1','bending2','cycling','lying','sitting','standing','walking']

# Load data directly using pandas
base_path = "../Data/AReM"
for activity in file_structure.keys():
    # Process test files
    for i in file_structure[activity]['test']:
        file_path = f"{base_path}/{activity}/dataset{i}.csv"
        try:
            df = pd.read_csv(file_path, skiprows=5, header=None, index_col=0)
            test_data.append(df.values)
            test_labels.append(f"{activity}/dataset{i}.csv")
        except FileNotFoundError:
            print(f"File not found: {file_path}")

    # Process train files
    for i in file_structure[activity]['train']:
        file_path = f"{base_path}/{activity}/dataset{i}.csv"
        try:
            df = pd.read_csv(file_path, skiprows=5, header=None, index_col=0)
            train_data.append(df.values)
            train_labels.append(f"{activity}/dataset{i}.csv")
        except FileNotFoundError:
            print(f"File not found: {file_path}")

# Feature extraction configuration
labels = ['min', 'max', 'mean', 'median', 'std', '1st quart', '3rd quart']
col_names = [f"{stat}_{sensor+1}" for sensor in range(6) for stat in labels]

# Function to extract features from data arrays
def extract_features(data_arrays, labels_list):
    features = defaultdict(list)
    for arr, label in zip(data_arrays, labels_list):
        # Calculate statistical measures for each column in the array        
        stats = [
            np.round(np.min(arr, axis=0), 2),
            np.round(np.max(arr, axis=0), 2),
            np.round(np.mean(arr, axis=0), 2),
            np.round(np.median(arr, axis=0), 2),
            np.round(np.std(arr, axis=0), 2),
            np.round(np.quantile(arr, 0.25, axis=0), 2),
            np.round(np.quantile(arr, 0.75, axis=0), 2)
        ]
        flattened = np.array(stats).T.flatten().tolist()
        features[label] = flattened
    return pd.DataFrame.from_dict(features, orient='index', columns=col_names).sort_index()

# Create feature dataframes
train_features = extract_features(train_data, train_labels)
test_features = extract_features(test_data, test_labels)

print("Training Features Shape:", train_features.shape)
print("Testing Features Shape:", test_features.shape)
```

### (c) Feature Extraction

##### (i) Research what types of time-domain features are usually used in time series classification and list them

```python

# Feature extraction configuration
labels = ['min', 'max', 'mean', 'median', 'std', '1st quart', '3rd quart']
col_names = [f"{stat}_{sensor+1}" for sensor in range(6) for stat in labels]

# Function to extract features from data arrays
def extract_features(data_arrays, labels_list):
    features = defaultdict(list)
    for arr, label in zip(data_arrays, labels_list):
        # Calculate statistical measures for each column in the array
        stats = [
            np.round(np.min(arr, axis=0), 2),
            np.round(np.max(arr, axis=0), 2),
            np.round(np.mean(arr, axis=0), 2),
            np.round(np.median(arr, axis=0), 2),
            np.round(np.std(arr, axis=0), 2),
            np.round(np.quantile(arr, 0.25, axis=0), 2),
            np.round(np.quantile(arr, 0.75, axis=0), 2)
        ]
        flattened = np.array(stats).T.flatten().tolist()
        features[label] = flattened
    return pd.DataFrame.from_dict(features, orient='index', columns=col_names).sort_index()

# Create feature dataframes
train_features = extract_features(train_data, train_labels)
test_features = extract_features(test_data, test_labels)

print("Training Features Shape:", train_features.shape)
print("Testing Features Shape:", test_features.shape)
```

#### (ii) Extract the time-domain features minimum, maximum, mean, median, standard deviation, first quartile, and third quartile for all of the 6 time series in each instance. You are free to normalize/standardize features or use them directly.


```python
# Add this after creating the features dataframes
pd.set_option('display.max_columns', None)  # Show all columns
pd.set_option('display.width', 1000)       # Set table width
pd.set_option('display.max_colwidth', 15)  # Column content width



# Combine the train_features and test_features tables
combined_features = pd.concat([train_features, test_features], axis=0).reset_index(drop=True)

# Add a new column 'Instance' starting from 1
combined_features.insert(0, 'Instance', range(1, len(combined_features) + 1))

# Print and display the combined table
print("\n" + "="*80)
print("FEATURES TABLE")
print("="*80)

# Style the table and hide the index
styled_table = combined_features.style.set_properties(**{
    'background-color': '#f8f9fa',
    'color': '#212529',
    'border': '1px solid #dee2e6'
}).hide()  # This hides the index

display(styled_table)
```

#### (iii) Estimate the standard deviation of each of the time-domain features you extracted from the data. Then, use Python’s bootstrapped or any other method to build a 90% bootsrap confidence interval for the standard deviation of each feature.

```python
# Compute standard deviations of each feature column
feature_std = combined_features.iloc[:, 1:].std()

# Function to bootstrap confidence intervals
def bootstrap_ci(data, num_samples=1000, ci=90):
    """Computes bootstrap confidence interval for standard deviation."""
    boot_samples = bs.bootstrap(data.values, stat_func=bs_stats.std, num_iterations=num_samples, alpha=(1 - ci / 100))
    return boot_samples.lower_bound, boot_samples.upper_bound

# Apply bootstrap CI to each feature column
bootstrap_results = {feature: bootstrap_ci(combined_features[feature]) for feature in combined_features.columns[1:]}

# Convert to DataFrame
ci_df = pd.DataFrame(bootstrap_results, index=['Lower Bound', 'Upper Bound']).T

# Combine standard deviations with confidence intervals
final_df = pd.DataFrame({'Std Dev': feature_std,
                         'Lower Bound': ci_df['Lower Bound'],
                         'Upper Bound': ci_df['Upper Bound']})

# Style the table with bar visualization
styled_final_df = final_df.style.bar(subset=['Std Dev'], align='mid', color=['#d65f5f', '#5fba7d'])

# Display the styled table
print("\nStandard Deviations and 90% Bootstrap Confidence Intervals:\n")
display(styled_final_df)
```

#### (iv) Use your judgement to select the three most important time-domain features(one option may be min, mean, and max)

```python
# Compute standard deviation for each feature (excluding 'Instance' column)
feature_std = combined_features.iloc[:, 1:].std()

# Select top 3 features with highest standard deviation
top_3_features = feature_std.nlargest(3)

# Print selected features
print("Top 3 Important Features (Based on Variability):")
print(top_3_features)
```

```python
import matplotlib.pyplot as plt

# Sort features by standard deviation (descending)
results_df_sorted = final_df.sort_values('Std Dev', ascending=False).reset_index()
results_df_sorted.rename(columns={'index': 'Feature'}, inplace=True)

# Create line graph visualization for all features (highlighting top 3)
plt.figure(figsize=(16, 8))

# Plot standard deviations with confidence intervals
plt.plot(results_df_sorted['Feature'], results_df_sorted['Std Dev'],
         marker='o', linestyle='--', color='silver', alpha=0.5,
         label='All Features')

# Highlight top 3 features
top3_features = results_df_sorted.head(3)
for idx, row in top3_features.iterrows():
    plt.plot(row['Feature'], row['Std Dev'],
             marker='*', markersize=12, color='#d62728',
             linestyle='None', label='Top 3' if idx == top3_features.index[0] else "")

# Add confidence interval range
plt.fill_between(results_df_sorted['Feature'],
                 results_df_sorted['Lower Bound'],
                 results_df_sorted['Upper Bound'],
                 color='skyblue', alpha=0.2)

# Formatting
plt.title('Feature Variability Spectrum (All Features)', fontsize=14, pad=20)
plt.xlabel('Features', fontsize=12)
plt.ylabel('Standard Deviation', fontsize=12)
plt.xticks(rotation=90, fontsize=8)
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# Corrected numerical display section
print("\nTop 3 Most Variable Features:")
display(
    top3_features.style
    .format({'Std Dev': '{:.3f}', 'Lower Bound': '{:.3f}', 'Upper Bound': '{:.3f}'})
    .set_caption("Top 3 Features by Variability")
    .background_gradient(subset=['Std Dev'], cmap='Oranges')
    .set_properties(**{
        'font-weight': 'bold',
        'background-color': '#fff7ec',
        'border': '1px solid #fee8c8'
    })
)
```

**1. I collect a set of data (n = 100 observations) containing a single predictor and a quantitative response. I then ft a linear regression model to the data, as well as a separate cubic regression, i.e. Y = β0 + β1X + β2X2 + β3X3 + ϵ.**

**(a) Linear Train**

The cubic model has more flexibility due to extra terms (X2,X3), allowing it to fit the data at least as well as the linear model. Thus, the cubic model's training RSS will be lower or equal to that of the linear model.

**(b) Linear Test**

Since the true relationship is linear, the cubic model may overfit the training data, capturing noise rather than the actual pattern. This results in higher or equal test RSS for the cubic model compared to the linear model.


**(c) Non Linear Train**

The cubic model is more flexible than the linear model, so it can capture more complex relationships in the data. Regardless of how far the true relationship is from linear, the cubic model will always fit the training data at least as well as the linear model. Therefore, the training RSS for the cubic model will be lower or equal to that of the linear model.

**(d) Non Linear Test**

If the true relationship is significantly nonlinear, the cubic model might provide a better fit, leading to a lower test RSS. However, if the cubic model captures too much noise, it may overfit, resulting in a higher test RSS. Since we do not know how far the true relationship is from linear, there is not enough information to determine which test RSS will be lower—it depends on the degree of nonlinearity and the extent of overfitting.
